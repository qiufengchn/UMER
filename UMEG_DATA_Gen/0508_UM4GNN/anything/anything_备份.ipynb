```python
import matplotlib.pyplot as plt
import random
import pandas as pd
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

# 设置随机种子，确保结果可复现
random.seed(42)

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体
plt.rcParams['axes.unicode_minus'] = False   # 解决负号显示问题

def calculate_energy_consumption(building):
    """
    根据建筑属性计算能源消耗。
    能耗公式：能耗 = 面积 * 层数 * 功能系数 * 年份系数
    """
    area = building["width"] * building["height"]
    floors = building["floors"]
    function_factor = {"residential": 1.0, "commercial": 1.5, "industrial": 2.0}[building["function"]]
    year_factor = 1.0 if building["year_built"] >= 2000 else 1.2  # 2000年后建筑更节能
    energy_consumption = area * floors * function_factor * year_factor
    return energy_consumption

def calculate_distance_to_main_road(building, roads):
    """
    计算建筑到最近主干道的距离
    """
    min_distance = float('inf')
    for road in roads:
        if road[2]["type"] == "main":
            start, end = road[0], road[1]
            # 计算建筑中心点到道路线段的距离
            building_x, building_y = building["center"]
            road_x1, road_y1 = start
            road_x2, road_y2 = end
            
            # 计算点到线段的距离
            if road_x1 == road_x2:  # 垂直道路
                distance = abs(building_x - road_x1)
            else:  # 水平道路
                distance = abs(building_y - road_y1)
            
            min_distance = min(min_distance, distance)
    
    # 如果没有主干道，设置一个默认较大距离
    if min_distance == float('inf'):
        min_distance = 10.0
        
    return min_distance

def calculate_accessibility(distance):
    """
    根据距离计算可达性（距离越近，可达性越高）
    """
    return 1 / (1 + distance)

def calculate_transportation_energy(building, accessibility):
    """
    计算交通能耗（可达性越低，交通能耗越高）
    """
    base_energy = building["energy_consumption"] * 0.1  # 基础交通能耗为建筑能耗的10%
    return base_energy * (2 - accessibility)  # 可达性越低，能耗越高

def generate_city_with_energy_and_accessibility(grid_size, road_probability=0.7, building_probability=0.5):
    """
    生成城市道路和建筑物的网格，并计算建筑物的能源消耗和可达性。
    """
    # 初始化道路和建筑物数据
    roads = []
    buildings = []
    road_traffic_energy = {}  # 用于存储每条道路的交通能耗

    # 遍历网格，生成横向和纵向道路
    for x in range(grid_size):
        for y in range(grid_size):
            # 横向道路
            if x < grid_size - 1 and random.random() < road_probability:
                road_type = "main" if random.random() > 0.5 else "secondary"
                roads.append(((x, y), (x + 1, y), {"width": random.uniform(0.1, 0.3), "type": road_type, "traffic_energy": 0}))
            # 纵向道路
            if y < grid_size - 1 and random.random() < road_probability:
                road_type = "main" if random.random() > 0.5 else "secondary"
                roads.append(((x, y), (x, y + 1), {"width": random.uniform(0.1, 0.3), "type": road_type, "traffic_energy": 0}))

    # 遍历道路，生成建筑物
    for road in roads:
        start, end, road_data = road
        mid_x = (start[0] + end[0]) / 2
        mid_y = (start[1] + end[1]) / 2

        if random.random() < building_probability:
            building = {
                "center": (mid_x + random.uniform(-0.3, 0.3), mid_y + random.uniform(-0.3, 0.3)),
                "width": random.uniform(0.2, 0.5),
                "height": random.uniform(0.2, 0.5),
                "floors": random.randint(1, 5),
                "function": random.choice(["residential", "commercial", "industrial"]),
                "year_built": random.randint(1950, 2025),
                "nearest_road": road  # 记录最近的道路
            }

            # 确保建筑物不与其他建筑物重叠
            overlap = False
            for existing_building in buildings:
                ex_center_x, ex_center_y = existing_building["center"]
                ex_width, ex_height = existing_building["width"], existing_building["height"]
                if abs(building["center"][0] - ex_center_x) < (building["width"] + ex_width) / 2 and \
                   abs(building["center"][1] - ex_center_y) < (building["height"] + ex_height) / 2:
                    overlap = True
                    break

            if not overlap:
                buildings.append(building)

    # 计算每栋建筑的能源消耗和交通能耗
    for building in buildings:
        building["energy_consumption"] = calculate_energy_consumption(building)
        building["distance_to_main_road"] = calculate_distance_to_main_road(building, roads)
        building["accessibility"] = calculate_accessibility(building["distance_to_main_road"])
        building["transportation_energy"] = calculate_transportation_energy(building, building["accessibility"])
        
        # 将交通能耗添加到最近道路的记录中
        road_id = f"{building['nearest_road'][0]}-{building['nearest_road'][1]}"
        if road_id not in road_traffic_energy:
            road_traffic_energy[road_id] = 0
        road_traffic_energy[road_id] += building["transportation_energy"]
        
        # 更新道路的交通能耗数据
        building["nearest_road"][2]["traffic_energy"] += building["transportation_energy"]

    # 将建筑物数据转换为表格
    building_data = [{
        "center": building["center"],
        "width": building["width"],
        "height": building["height"],
        "floors": building["floors"],
        "function": building["function"],
        "year_built": building["year_built"],
        "energy_consumption": building["energy_consumption"],
        "distance_to_main_road": building["distance_to_main_road"],
        "accessibility": building["accessibility"],
        "transportation_energy": building["transportation_energy"]
    } for building in buildings]
    
    # 将道路数据转换为表格
    road_data = [{
        "start": road[0], 
        "end": road[1], 
        "width": road[2]["width"], 
        "type": road[2]["type"], 
        "traffic_energy": road[2]["traffic_energy"]
    } for road in roads]

    building_df = pd.DataFrame(building_data)
    road_df = pd.DataFrame(road_data)
    
    print("\n建筑物数据：")
    print(building_df)
    
    print("\n道路交通能耗数据：")
    print(road_df)

    # 创建颜色映射
    energy_values = [b["energy_consumption"] for b in buildings]
    min_energy = min(energy_values) if energy_values else 0
    max_energy = max(energy_values) if energy_values else 1
    
    # 创建颜色映射（从浅到深）
    cmap = LinearSegmentedColormap.from_list('energy_cmap', ['#FFE4E1', '#FF0000'])

    # 绘制城市网格 - 能耗图
    fig, ax = plt.subplots(figsize=(12, 10))

    # 绘制道路
    for road in roads:
        start, end, road_data = road
        color = "black" if road_data["type"] == "main" else "gray"
        ax.plot([start[0], end[0]], [start[1], end[1]], color=color, linewidth=road_data["width"] * 10, alpha=0.8)

    # 绘制建筑物
    for building in buildings:
        center_x, center_y = building["center"]
        width = building["width"]
        height = building["height"]
        
        # 根据能耗值设置颜色
        normalized_energy = (building["energy_consumption"] - min_energy) / (max_energy - min_energy) if max_energy != min_energy else 0.5
        color = cmap(normalized_energy)
        
        ax.add_patch(plt.Rectangle(
            (center_x - width / 2, center_y - height / 2),
            width,
            height,
            color=color,
            alpha=0.6
        ))

    # 添加图例
    main_road = plt.Line2D([0], [0], color='black', lw=4, label='主干道')
    secondary_road = plt.Line2D([0], [0], color='gray', lw=4, label='次干道')
    ax.legend(handles=[main_road, secondary_road], loc="upper right")

    # 添加颜色条
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(min_energy, max_energy))
    plt.colorbar(sm, ax=ax, label='建筑能源消耗')

    ax.set_title(f"城市建筑能耗分布图 ({grid_size}x{grid_size})")
    ax.axis("equal")
    plt.tight_layout()
    plt.show()

    # 绘制交通能耗图
    fig, ax = plt.subplots(figsize=(12, 10))
    
    # 绘制道路
    for road in roads:
        start, end, road_data = road
        color = "black" if road_data["type"] == "main" else "gray"
        ax.plot([start[0], end[0]], [start[1], end[1]], color=color, linewidth=road_data["width"] * 10, alpha=0.8)

    # 绘制建筑物 - 根据交通能耗着色
    trans_values = [b["transportation_energy"] for b in buildings]
    min_trans = min(trans_values) if trans_values else 0
    max_trans = max(trans_values) if trans_values else 1
    trans_cmap = LinearSegmentedColormap.from_list('trans_cmap', ['#90EE90', '#006400'])  # 从浅绿到深绿
    
    for building in buildings:
        center_x, center_y = building["center"]
        width = building["width"]
        height = building["height"]
        
        # 根据交通能耗设置颜色
        normalized_trans = (building["transportation_energy"] - min_trans) / (max_trans - min_trans) if max_trans != min_trans else 0.5
        color = trans_cmap(normalized_trans)
        
        ax.add_patch(plt.Rectangle(
            (center_x - width / 2, center_y - height / 2),
            width,
            height,
            color=color,
            alpha=0.6
        ))

    # 添加图例
    main_road = plt.Line2D([0], [0], color='black', lw=4, label='主干道')
    secondary_road = plt.Line2D([0], [0], color='gray', lw=4, label='次干道')
    ax.legend(handles=[main_road, secondary_road], loc="upper right")
    
    # 添加颜色条
    sm = plt.cm.ScalarMappable(cmap=trans_cmap, norm=plt.Normalize(min_trans, max_trans))
    plt.colorbar(sm, ax=ax, label='建筑交通能耗')

    ax.set_title(f"城市建筑交通能耗分布图 ({grid_size}x{grid_size})")
    ax.axis("equal")
    plt.tight_layout()
    plt.show()

# 调用函数生成城市并计算能耗和可达性
generate_city_with_energy_and_accessibility(grid_size=10, road_probability=0.7, building_probability=0.5)
```

```python
    # ==== 可视化 1: 建筑能耗分布 ====
    fig, ax = plt.subplots(figsize=(12, 10))
    # 绘制道路
    for road in roads:
        start, end, road_data = road
        color = "black" if road_data["type"] == "main" else "gray"
        ax.plot([start[0], end[0]], [start[1], end[1]], color=color, 
                linewidth=road_data["width"] * 10, alpha=0.8)
    
    # 绘制建筑物，根据能耗着色
    for building in buildings:
        center_x, center_y = building["center"]
        width = building["width"]
        height = building["height"]
        
        # 能耗归一化与着色
        normalized_energy = ((building["energy_consumption"] - min_energy) / 
                             (max_energy - min_energy)) if max_energy != min_energy else 0.5
        color = building_cmap(normalized_energy)
        
        ax.add_patch(plt.Rectangle(
            (center_x - width / 2, center_y - height / 2),
            width, height, color=color, alpha=0.7
        ))
    
    # 添加图例
    main_road = plt.Line2D([0], [0], color='black', lw=4, label='主干道')
    secondary_road = plt.Line2D([0], [0], color='gray', lw=4, label='次干道')
    ax.legend(handles=[main_road, secondary_road], loc="upper right")
    
    # 添加颜色条
    sm = plt.cm.ScalarMappable(cmap=building_cmap, norm=plt.Normalize(min_energy, max_energy))
    sm.set_array([])  # 设置空数组以避免警告
    plt.colorbar(sm, ax=ax, label='建筑能源消耗')  # 添加ax参数
    
    ax.set_title("城市建筑能耗分布图")
    ax.axis("equal")
    plt.tight_layout()
    plt.show()
    
    # ==== 可视化 2: 交通能耗分布 ====
    fig, ax = plt.subplots(figsize=(12, 10))
    # 绘制道路
    for road in roads:
        start, end, road_data = road
        color = "black" if road_data["type"] == "main" else "gray"
        ax.plot([start[0], end[0]], [start[1], end[1]], color=color, 
                linewidth=road_data["width"] * 10, alpha=0.8)
    
    # 绘制建筑物，根据交通能耗着色
    for building in buildings:
        center_x, center_y = building["center"]
        width = building["width"]
        height = building["height"]
        
        # 交通能耗归一化与着色
        normalized_trans = ((building["transportation_energy"] - min_trans) / 
                           (max_trans - min_trans)) if max_trans != min_trans else 0.5
        color = trans_cmap(normalized_trans)
        
        ax.add_patch(plt.Rectangle(
            (center_x - width / 2, center_y - height / 2),
            width, height, color=color, alpha=0.7
        ))
    
    # 添加图例
    main_road = plt.Line2D([0], [0], color='black', lw=4, label='主干道')
    secondary_road = plt.Line2D([0], [0], color='gray', lw=4, label='次干道')
    ax.legend(handles=[main_road, secondary_road], loc="upper right")
    
    # 添加颜色条
    sm = plt.cm.ScalarMappable(cmap=trans_cmap, norm=plt.Normalize(min_trans, max_trans))
    sm.set_array([])
    plt.colorbar(sm, ax=ax, label='建筑交通能耗')
    
    ax.set_title("城市建筑交通能耗分布图")
    ax.axis("equal")
    plt.tight_layout()
    plt.show()
    
    # ==== 可视化 3: 城市区域划分 ====
    fig, ax = plt.subplots(figsize=(12, 10))
    # 绘制道路
    for road in roads:
        start, end, road_data = road
        color = "black" if road_data["type"] == "main" else "gray"
        ax.plot([start[0], end[0]], [start[1], end[1]], color=color, 
                linewidth=road_data["width"] * 10, alpha=0.5)
    
    # 绘制建筑物，根据区域着色
    for building in buildings:
        center_x, center_y = building["center"]
        width = building["width"]
        height = building["height"]
        
        district_id = building["district"]
        color = district_colors[district_id % len(district_colors)]
        
        ax.add_patch(plt.Rectangle(
            (center_x - width / 2, center_y - height / 2),
            width, height, color=color, alpha=0.7
        ))
    
    # 添加图例
    district_patches = []
    for i in range(len(district_data)):
        district_patches.append(mpatches.Patch(
            color=district_colors[i % len(district_colors)], 
            label=f'区域 {i}'
        ))
    
    ax.legend(handles=district_patches, loc="upper right")
    ax.set_title("城市区域划分图")
    ax.axis("equal")
    plt.tight_layout()
    plt.show()
```
